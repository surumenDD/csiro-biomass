```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ==== 入力 ====
TRAIN_CSV = Path("../../input/train.csv")
FOLD_CSV  = Path("../../artifacts/folds_opt.csv")  # fold割当（row_id なし版）

# ==== 列定義（make_folds_fixed_opt.py と揃える）====
META_COLS = [
    "sample_id_prefix",
    "image_path",
    "Sampling_Date",
    "State",
    "Species",
    "Pre_GSHH_NDVI",
    "Height_Ave_cm",
]
TARGETS = ["Dry_Green_g", "Dry_Dead_g", "Dry_Clover_g"]

# =========================
# 1) train.csv -> wide
# =========================
df = pd.read_csv(TRAIN_CSV)

sp = df["sample_id"].astype(str).str.split("__", n=1, expand=True)
df["sample_id_prefix"] = sp[0]

wide = (
    df.pivot_table(
        index=META_COLS,
        columns="target_name",
        values="target",
        aggfunc="first",
    )
    .reset_index()
)
wide.columns.name = None

# =========================
# 2) fold を付与（row_idを使わず META_COLS で結合）
#   folds_opt.csv 側に META_COLS + fold が入っている前提
# =========================
fold_df = pd.read_csv(FOLD_CSV)

need = META_COLS + ["fold"]
miss = [c for c in need if c not in fold_df.columns]
if miss:
    raise ValueError(
        "fold_csv に必要な列がありません: "
        + ", ".join(miss)
        + "\nfold_csv には META_COLS + fold を入れてください。"
    )

# 型ズレで join 失敗しやすい列は、同じ型に寄せる（文字列化）
# ※数値は float 表現ズレが起こりうるので、ここでは文字列化せず、そのままにする。
#   ただし Sampling_Date / State / Species / image_path / sample_id_prefix は文字列推奨。
str_cols = ["sample_id_prefix", "image_path", "Sampling_Date", "State", "Species"]
for c in str_cols:
    wide[c] = wide[c].astype(str)
    fold_df[c] = fold_df[c].astype(str)

# fold を付与
wide = wide.merge(fold_df[need], on=META_COLS, how="inner")

# 結合漏れチェック（fold_df 側を基準に作っているなら通常 357 行になるはず）
print("=== merge check ===")
print(f"wide rows after merge: {len(wide)}")

# =========================
# 3) foldサイズ確認
# =========================
vc = wide["fold"].value_counts().sort_index()
print("\n=== fold sizes (rows) ===")
for f in sorted(vc.index.tolist()):
    print(f"fold={f}: n_rows={int(vc.loc[f])}")
print(f"size_ratio_rows={float(vc.max()/max(1,int(vc.min()))):.6f}")

# =========================
# 4) テーブル（コピーしやすい TSV を同時に出す）
#   - stats（min/max/分位）
#   - 0率 / high率 / p95/p99 / top10% mean 等
# =========================
def make_tables(df_in: pd.DataFrame, targets: list[str], high_q: float = 0.90):
    # 全体の high 閾値（fold間で比較するため固定）
    thr = df_in[targets].apply(pd.to_numeric, errors="coerce").quantile(high_q)

    rows_stats = []
    rows_rich = []

    def one_block(tag, sub):
        nonlocal rows_stats, rows_rich

        for t in targets:
            x = pd.to_numeric(sub[t], errors="coerce").dropna()
            if len(x) == 0:
                continue

            # basic stats
            rows_stats.append({
                "fold": tag, "target": t,
                "count": int(x.shape[0]),
                "mean": float(x.mean()),
                "std": float(x.std(ddof=0)),
                "min": float(x.min()),
                "q25": float(x.quantile(0.25)),
                "median": float(x.quantile(0.50)),
                "q75": float(x.quantile(0.75)),
                "max": float(x.max()),
            })

            # richer diagnostics（尾の偏りを見る）
            p95 = float(x.quantile(0.95))
            p99 = float(x.quantile(0.99))
            thr_t = float(thr[t])

            k10 = max(1, int(np.ceil(len(x) * 0.10)))
            k05 = max(1, int(np.ceil(len(x) * 0.05)))
            xs = np.sort(x.values)

            rows_rich.append({
                "fold": tag, "target": t,
                "count": int(len(x)),
                "zero_rate": float((x == 0).mean()),
                "high_q": high_q,
                "high_thr": thr_t,
                "high_rate": float((x >= thr_t).mean()),
                "mean": float(x.mean()),
                "q75": float(x.quantile(0.75)),
                "p95": p95,
                "p99": p99,
                "top10%_mean": float(xs[-k10:].mean()),
                "top5%_mean": float(xs[-k05:].mean()),
                "max": float(x.max()),
            })

    # all
    one_block("all", df_in)
    # folds
    for f in sorted(df_in["fold"].unique().tolist()):
        one_block(int(f), df_in[df_in["fold"] == f])

    stats = pd.DataFrame(rows_stats)
    rich = pd.DataFrame(rows_rich)
    return stats, rich, thr

stats, rich, thr = make_tables(wide, TARGETS, high_q=0.90)

# 並びを揃える
def _fold_key(v):
    return -1 if str(v) == "all" else int(v)

stats = stats.assign(_foldk=stats["fold"].map(_fold_key)).sort_values(["target","_foldk"]).drop(columns="_foldk")
rich  = rich.assign(_foldk=rich["fold"].map(_fold_key)).sort_values(["target","_foldk"]).drop(columns="_foldk")

# Notebook上で見やすく表示
display(stats)
display(rich)

# ===== ここが「一括コピー」用の出力 =====
# TSV はそのままスプレッドシートに貼れる
print("\n=== COPY-PASTE TSV: stats ===")
print(stats.to_csv(sep="\t", index=False))

print("\n=== COPY-PASTE TSV: rich ===")
print(rich.to_csv(sep="\t", index=False))

# =========================
# 5) 可視化：q99 クリップ hist（all vs fold）
# =========================
q99 = wide[TARGETS].apply(pd.to_numeric, errors="coerce").quantile(0.99)

for t in TARGETS:
    xmax = float(q99[t]) if np.isfinite(q99[t]) else None

    plt.figure()
    x_all = pd.to_numeric(wide[t], errors="coerce").dropna().values
    if xmax is not None:
        x_all = x_all[x_all <= xmax]
    plt.hist(x_all, bins=40, density=True, histtype="step", linewidth=2, label="all")

    for f in sorted(wide["fold"].unique().tolist()):
        sub = wide[wide["fold"] == f]
        x = pd.to_numeric(sub[t], errors="coerce").dropna().values
        if xmax is not None:
            x = x[x <= xmax]
        plt.hist(x, bins=40, density=True, histtype="step", linewidth=1, label=f"fold{f}")

    plt.title(f"{t}  density hist (<= q99={xmax:.4f})")
    plt.xlabel(t)
    plt.ylabel("density")
    plt.legend()
    plt.show()

# =========================
# 6) log1p hist（裾 + 0 多さ確認）
# =========================
for t in TARGETS:
    plt.figure()
    x_all = pd.to_numeric(wide[t], errors="coerce").dropna().values
    plt.hist(np.log1p(x_all), bins=40, density=True, histtype="step", linewidth=2, label="all")

    for f in sorted(wide["fold"].unique().tolist()):
        sub = wide[wide["fold"] == f]
        x = pd.to_numeric(sub[t], errors="coerce").dropna().values
        plt.hist(np.log1p(x), bins=40, density=True, histtype="step", linewidth=1, label=f"fold{f}")

    plt.title(f"{t}  density hist (log1p)")
    plt.xlabel(f"log1p({t})")
    plt.ylabel("density")
    plt.legend()
    plt.show()
```
